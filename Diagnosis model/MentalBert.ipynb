```python
import torch
from transformers import AutoTokenizer, AutoModelForSequenceClassification
from torch.utils.data import DataLoader
from datasets import Dataset
import numpy as np
```

```python
domain_data.sample(frac=1,random_state=42)
domain_to_test=domain_data[:10000]
```

```python

domain_to_test['label']=1
```

```python
dataset = Dataset.from_pandas(domain_to_test)

def tokenize(batch):
    return tokenizer(
        batch['other_posts'],
        truncation=True,
        padding="max_length",
        max_length=512
    )

dataset = dataset.map(tokenize, batched=True)
dataset.set_format(
    type="torch",
    columns=["input_ids", "attention_mask", "label"]
)

test_loader = DataLoader(dataset, batch_size=32, shuffle=False)
```

```python
from torch.nn.functional import softmax

total_loss = 0
all_preds = []
all_probs = []
all_labels = []

criterion = torch.nn.CrossEntropyLoss()

with torch.no_grad():
    for batch in test_loader:
        input_ids = batch["input_ids"].to(device)
        attention_mask = batch["attention_mask"].to(device)
        labels = batch["label"].to(device)

        outputs = model(
            input_ids=input_ids,
            attention_mask=attention_mask
        )

        logits = outputs.logits
        loss = criterion(logits, labels)

        probs = softmax(logits, dim=1)
        preds = torch.argmax(probs, dim=1)

        total_loss += loss.item()

        all_preds.extend(preds.cpu().numpy())
        all_probs.extend(probs[:, 1].cpu().numpy())  # prob of mental illness
        all_labels.extend(labels.cpu().numpy())
```

```python
from sklearn.metrics import classification_report
all_preds = np.array(all_preds)
all_labels = np.array(all_labels)

accuracy = (all_preds == all_labels).mean()

false_positives = (all_preds == 1).sum()
false_positive_rate = false_positives / len(all_labels)

print(f"Test loss: {total_loss / len(test_loader):.4f}")
print(f"Accuracy (all normal): {accuracy:.4f}")
print(f"False Positive Rate: {false_positive_rate:.4f}")
print("Prediction distribution:", np.bincount(all_preds))
print('='*50)
print(classification_report(all_labels,all_preds))
```
