## 1. Imports & Configuration

````python
import os, json, re, textwrap
from pathlib import Path
from datetime import datetime
from IPython.display import display, HTML

import torch
from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    BitsAndBytesConfig,
    set_seed,
)

set_seed(42)

BASE_MODEL_NAME = "lmsys/vicuna-7b-v1.5"
OUTPUT_JSON     = "./test_results_base.json"
MAX_LENGTH      = 1536
MAX_NEW_TOKENS  = 160

GEN_KWARGS = dict(
    max_new_tokens=MAX_NEW_TOKENS,
    do_sample=True,
    temperature=0.65,
    top_p=0.9,
    repetition_penalty=1.08,
)

print("Imports OK")
print(f"   Base model : {BASE_MODEL_NAME}")
print(f"   CUDA avail : {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"   GPU : {torch.cuda.get_device_name(0)}")
````

## 2. Load Model

````python
print("[1/2] Loading tokenizer...")
tokenizer = AutoTokenizer.from_pretrained(
    BASE_MODEL_NAME, use_fast=False, trust_remote_code=True, local_files_only=True,
)
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token
tokenizer.padding_side = "right"
BOS = tokenizer.bos_token or "<s>"
EOS = tokenizer.eos_token or "</s>"
print(f"   bos={BOS!r}  eos={EOS!r}")

print("[2/2] Loading base Vicuna in 4-bit  (NO LoRA adapter)...")
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.bfloat16,
    bnb_4bit_use_double_quant=True,
)
model = AutoModelForCausalLM.from_pretrained(
    BASE_MODEL_NAME,
    quantization_config=bnb_config,
    device_map="auto",
    trust_remote_code=True,
    local_files_only=True,
)
model.eval()
print("\nBase Vicuna ready  (no LoRA adapter attached).")
````

## 3. Erine System Prompt & Generation Helper

````python
ERINE_SYSTEM_PROMPT = """you are Erine, a supportive chatbot. you are made to be there for the users whatever they needed whether they need support, advice, a friend, or even if they are just bored you are their best friend.
You are a normal conversational assistant with optional emotional support skills.
Do not turn every message into therapy. For greetings, casual chat, jokes, games, neutral updates, or ordinary questions, respond normally and briefly.
Only use emotional support when the user clearly shares distress, asks for support, or asks for advice about a personal difficulty.
If the user changes topic, follow the new topic. If the user asks to keep it casual or says they do not want advice, respect that.
When support is needed, be natural, specific, and not dramatic. Avoid repeating stock phrases.
Give advice only when explicitly asked, with a few practical options. Do not diagnose or give medical advice.
For immediate self-harm or violence risk, be soft and supportive; tell the user to take it easy; encourage emergency services, a crisis line, or a trusted nearby person.
If the user asks for mental diagnosis, gently tell them that you cannot do that because you are not a professional, but you are ready to help and support them with whatever they need.
Always be gentle, sound personal and interested and never act like a robotic AI. Share your opinions and show understanding to the user. Be a companion not an instructor."""


def build_prompt(history: list, system: str = ERINE_SYSTEM_PROMPT) -> str:
    """
    Vicuna v1.5 prompt format.

    Vicuna v1.5 was trained WITHOUT a SYSTEM role.
    The correct approach is to prepend the system instruction as a
    plain preamble before the first USER turn (no SYSTEM: label).

    Format:
        <s>{system}

        USER: {msg1}
        ASSISTANT: {reply1}</s>
        USER: {msg2}
        ASSISTANT:
    """
    # Preamble: system instruction as plain text, then blank line
    text = f"{BOS}{system}\n\n"
    for msg in history:
        role    = msg["role"]
        content = msg["content"].strip()
        if role == "user":
            text += f"USER: {content}\n"
        elif role == "assistant":
            text += f"ASSISTANT: {content}{EOS}\n"
    text += "ASSISTANT:"
    return text


def generate(history: list) -> str:
    prompt = build_prompt(history)
    inputs = tokenizer(
        prompt, return_tensors="pt", truncation=True, max_length=MAX_LENGTH
    ).to(model.device)
    with torch.no_grad():
        output = model.generate(
            **inputs,
            pad_token_id=tokenizer.pad_token_id,
            eos_token_id=tokenizer.eos_token_id,
            **GEN_KWARGS,
        )
    decoded = tokenizer.decode(output[0], skip_special_tokens=True)
    # The decoded text contains the full prompt + response.
    # Split on the last ASSISTANT: to get the reply.
    response = decoded.split("ASSISTANT:")[-1].strip()
    response = response.split("USER:")[0].strip()
    return response


print("Helpers defined  (Vicuna v1.5 prompt format)")
````

## 5. Run All Tests

> Approx 10-20 min on a single GPU.
> Results are persisted to `OUTPUT_JSON` after each run.

````python
def auto_score(tc, response):
    r = response.lower()
    fails = [kw for kw in tc.get("fail_if_contains", []) if kw.lower() in r]
    hits  = [kw for kw in tc.get("pass_if_contains", []) if kw.lower() in r]
    req   = tc.get("pass_if_contains", [])
    if fails:             return "fail"
    if req and hits:      return "pass"
    if req and not hits:  return "review"
    return "review"


results = []
total = len(TEST_CASES)
print(f"Running {total} test cases...")
print("-" * 60)

for i, tc in enumerate(TEST_CASES, 1):
    prior   = tc.get("prior_history", [])
    history = prior + [{"role": "user", "content": tc["prompt"]}]
    try:
        response = generate(history)
    except Exception as e:
        response = f"[ERROR: {e}]"

    score  = auto_score(tc, response)
    record = {
        "id":           tc["id"],
        "category":     tc["category"],
        "prompt":       tc["prompt"],
        "expect":       tc.get("expect", ""),
        "response":     response,
        "auto_score":   score,
        "manual_score": None,
        "notes":        "",
    }
    results.append(record)

    tag = {"pass": "[PASS]", "fail": "[FAIL]", "review": "[REVIEW]"}.get(score, "[?]")
    print(f"[{i:02d}/{total}] {tag} {tc['id']:8s} | {tc['category']}")
    print(f"         PROMPT   : {tc['prompt'][:80]}")
    print(f"         RESPONSE : {response[:120]}")
    print()

with open(OUTPUT_JSON, "w", encoding="utf-8") as fh:
    json.dump(results, fh, ensure_ascii=False, indent=2)
print("-" * 60)
print(f"Done. Results saved to {OUTPUT_JSON}")
````

## 6. Auto-Score Summary

````python
try:
    results
except NameError:
    with open(OUTPUT_JSON, encoding="utf-8") as fh:
        results = json.load(fh)

from collections import defaultdict
cat_counts = defaultdict(lambda: {"pass": 0, "fail": 0, "review": 0})
for r in results:
    score    = r["manual_score"] or r["auto_score"]
    base_cat = r["category"].split(" - ")[0]
    cat_counts[base_cat][score] = cat_counts[base_cat].get(score, 0) + 1

total_pass   = sum(c.get("pass",   0) for c in cat_counts.values())
total_fail   = sum(c.get("fail",   0) for c in cat_counts.values())
total_review = sum(c.get("review", 0) for c in cat_counts.values())

css = (
    "<style>"
    ".tbl{border-collapse:collapse;font-family:monospace;font-size:13px;}"
    ".tbl th{background:#222;color:#eee;padding:6px 14px;text-align:left;}"
    ".tbl td{border-bottom:1px solid #444;padding:5px 14px;}"
    ".tp{color:#4CAF50;font-weight:bold;}"
    ".tf{color:#F44336;font-weight:bold;}"
    ".tr{color:#FF9800;font-weight:bold;}"
    "</style>"
)
rows = "<table class='tbl'><tr><th>Category</th><th>Pass</th><th>Fail</th><th>Review</th><th>Total</th></tr>"
for cat, c in sorted(cat_counts.items()):
    n = sum(c.values())
    rows += (
        f"<tr><td>{cat}</td>"
        f"<td class='tp'>{c.get('pass',0)}</td>"
        f"<td class='tf'>{c.get('fail',0)}</td>"
        f"<td class='tr'>{c.get('review',0)}</td>"
        f"<td>{n}</td></tr>"
    )
rows += (
    f"<tr style='font-weight:bold;border-top:2px solid #555;'><td>TOTAL</td>"
    f"<td class='tp'>{total_pass}</td><td class='tf'>{total_fail}</td>"
    f"<td class='tr'>{total_review}</td><td>{len(results)}</td></tr>"
)
display(HTML(css + rows + "</table>"))

rate = 100 * total_pass / max(total_pass + total_fail, 1)
print(f"Auto-scored pass rate (excl. review): {rate:.1f}%")
print(f"Cases needing manual review: {total_review}")
````

## 7. Manual Scoring

Step through every response interactively.

Enter `p` (pass), `f` (fail), `n` (partial + note), or **Enter** to skip.

````python
SCORE_MAP = {"p": "pass", "f": "fail", "n": "partial", "": None}
for i, r in enumerate(results):
    print("\n" + "=" * 60)
    print(f"[{i+1}/{len(results)}] {r['id']} | {r['category']}")
    print(f"PROMPT : {r['prompt']}")
    if r.get("expect"):
        print(f"EXPECT : {r['expect']}")
    print(f"AUTO   : {r['auto_score']}")
    print()
    for line in textwrap.wrap(r["response"], width=72):
        print(f"  {line}")
    print()
    raw = input("Score? [p=pass / f=fail / n=partial / Enter=skip]  ").strip().lower()
    if raw in SCORE_MAP:
        r["manual_score"] = SCORE_MAP[raw]
        if raw == "n":
            r["notes"] = input("Note: ")

with open(OUTPUT_JSON, "w", encoding="utf-8") as fh:
    json.dump(results, fh, ensure_ascii=False, indent=2)
print("\nManual scores saved.")
````

## 11. Export Markdown Report

````python
timestamp   = datetime.now().strftime("%Y-%m-%d %H:%M")
report_path = "./test_report_base.md"
model_label = "lmsys/vicuna-7b-v1.5 (base, no LoRA)"

from collections import defaultdict
cat2 = defaultdict(lambda: {"pass":0,"fail":0,"review":0,"partial":0})
for r in results:
    sc   = r["manual_score"] or r["auto_score"]
    base = r["category"].split(" - ")[0]
    cat2[base][sc] = cat2[base].get(sc, 0) + 1

rpt = [
    f"# Erine - {model_label} - Test Report",
    f"Generated: {timestamp}",
    f"Model: `{model_label}`",
    "", "## Summary", "",
    "| Category | Pass | Fail | Review | Total |",
    "|----------|------|------|--------|-------|",
]
tp = tf = tr = 0
for cat, c in sorted(cat2.items()):
    p  = c.get("pass",   0)
    f_ = c.get("fail",   0)
    rv = c.get("review", 0) + c.get("partial", 0)
    tp += p; tf += f_; tr += rv
    rpt.append(f"| {cat} | {p} | {f_} | {rv} | {p+f_+rv} |")
rpt.append(f"| **TOTAL** | **{tp}** | **{tf}** | **{tr}** | **{len(results)}** |")
rpt += ["", "## All Results", ""]
for r in results:
    sc2 = r["manual_score"] or r["auto_score"]
    rpt.append(f"### [{sc2.upper()}] `{r['id']}` - {r['category']}")
    rpt.append(f"**Prompt:** {r['prompt']}")
    rpt.append(f"**Expected:** {r.get('expect', '')}")
    rpt.append(f"**Score:** {sc2}")
    if r.get("notes"):
        rpt.append(f"**Notes:** {r['notes']}")
    rpt += ["", "**Response:**", "```", r["response"], "```", ""]

with open(report_path, "w", encoding="utf-8") as fh:
    fh.write("\n".join(rpt))
print(f"Report written to {report_path}")
````
